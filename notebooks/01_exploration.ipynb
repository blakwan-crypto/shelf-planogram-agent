# 01 — Exploration
What do our sample shelves look like, and what does the detector see?

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

imgs = sorted(Path('../data/sample').glob('*.jpg'))
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, p in zip(axes.flat, imgs):
    ax.imshow(Image.open(p)); ax.set_title(p.name); ax.axis('off')
plt.tight_layout()
```

```python
# Detection counts + confidence distribution per image
import sys; sys.path.insert(0, '..')
from tools.detector import ProductDetector
det = ProductDetector(weights='../models/sku110k-yolo11-n640.pt')
for p in imgs:
    ds = det.detect(str(p))
    print(f'{p.name}: {len(ds)} boxes, mean conf {sum(d.confidence for d in ds)/max(len(ds),1):.2f}')
```
